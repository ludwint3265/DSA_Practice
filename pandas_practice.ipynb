```python
import pandas as pd
# import the pandas library 
```

```python
df = pd.DataFrame(
   {
      "Name": [
         "S, Joseph",
         "P, Marlon",
         "J, Bryan",
         "P, Lisbeth",
         "M, Chris",
         "T, Ludwin",
      ],
      "Age": [0, 19, 20, 21, 20, 21],
      "Height": [6.2, 5.8, 5.6, 5.3, 5.10, 5.3],
      "Sex": ["male", "male", "male", "female", "male", "male"]
   }
)
# DataFrames have columns, and a label for each column. They also have 0-indexed
# rows, and these indices can be accessed in different ways. This is manual instantiation.
# When using a Python dictionary of lists, keys = column headers, and the values
# in each list as the columns

print(df)
# Each column in a DataFrame is a Series
```

```python
# Print only the age column of the existing DataFrame; these columns are each 
# Pandas Series. A Series has a name and a data type, so you aren't able to 
# make a Series with multiple data types. If there are multiple data types in 
# a Series, Pandas will resolve this by trying to make them all the same data type.
# If they are numerical (int and float), all will change to float. If not, 
# they will be of object type.
print(df["Age"])
```

```python
# A Pandas Series
my_heights = pd.Series([12,2,4,23,40], name = "Heights")
print(my_heights)
```

```python
# DataFrames are two-dimensional data structures that can store
# data of different types (characters, floating point values)
# in columns. It's similar to a spreadsheet, or a SQL table.

# Each column in a DataFrame is a Series
print(df["Age"])
```

```python
# Get the maximum age of the Age Series
print(df["Age"].max())
```

```python
my_heights.max()
```

```python
# Pandas provies methods that can be applied to DataFrames
# or Series. For example, the describe() method shows 
# basic numerical data of a column. It can display the stats 
# of an individual Series, or the numerical stats of the 
# columns of a DataFrame. This method returns a Pandas DataFrame or 
# Series, depending on the number of columns returned.
df.describe()

```

```python
# How to read and write tabular data
#
# Pandas has the read_csv() function to read data 
# stored in a csv. Pandas supports many file formats to read,
# including excel, sql, json, parquet, all beginning with 
# read_*, depending on what extension is being read.

titanic = pd.read_csv("titanic.csv")
titanic
```

```python
# To see the first 8 rows:
titanic.head(8)
# The .head(N) method is used to see the first N rows of a
# DataFrame, from the top down, of the data (not including the column headers)
```

```python
# The .tail(N) method returns the last N rows of the DataFrame / Series it is called on.

titanic.tail(10)
```

```python
# We can check on how pandas interpreted each of the column 
# data types by requesting the pandas dtypes attributes:

titanic.dtypes

# We can see that the data types in this DataFrame are integers
# (int64), floats (float64), and strings (object)

# No parentheses are used to get a Series or DataFrame dtypes,
# because dtypes is an attribute of both. We are only returning the specific
# attribute dtypes, by accessing it directly on the object. We needed the
# parentheses on the previous method, because the previous method is defined as a function
# that takes an action, e.g. getting the maximum numerical value from a DataFrame / Series
```
