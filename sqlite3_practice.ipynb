<strong><h1>Sqlite3</h1></strong>

sqlite3 is a Python module that provides database (SQL) functionality without spending time and resources on spinning up an entirely new database.

```python
# starter code
import sqlite3

# sample Employee class
class Employee:
   def __init__(self, first, last, pay):
      self.first = first
      self.last = last
      self.pay = pay

   @property
   def email(self):
      return '{}.{}@email.com'.format(self.first, self.last)
   
   @property 
   def fullname(self):
      return '{} {}'.format(self.first, self.last)

   def __repr__(self):
      return "Employee('{}', '{}', {})".format(self.first, self.last, self.pay)
```

```python
# can create a database in memory like so:
conn1 = sqlite3.connect(':memory:')
print(conn1)

# in this format, the connection closes automatically once the code ends, since the 'database' lives in RAM
```

```python
# can also create a file to use as database
# this ended up creating a new file that will act as a database

conn = sqlite3.connect('employee.db')
# using triple quotes (docstring) which allows us to write a 
# multi-line string w/o using special line breaks
c = conn.cursor() # create a cursor to execute SQL commands
res = c.execute("""CREATE TABLE employees (
            first text,
            last text,
            pay integer
            )""")
# only wrote and ran this once, as it will try to create the same 
# table in the same database file that was made by the initial connect() call

conn.commit()

conn.close()
# Values stored in SQLITE database / manipulated by the database
# engine
# has one of the following storage classes:
# NULL: The value is a NULL value
# INTEGER: The value is a signed integer, stored in 1,2,3,4,6,
# or 8 bytes 
# depending on the magnitude of the value
# REAL: The value is a floating point value, stored as an 
# 8-byte IEEF floating point number
# TEXT: The value is a text string, stored using the database
# encoding (UTF-8, UTF-16BE or UTF-16LE)
# BLOB: The value is a blob of data, stored exactly as it was put
```
