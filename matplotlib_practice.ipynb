Matplotlib is a Python library used to plot graphs.

```python
# Minimal plot example:
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2*np.pi, 200)
y = np.sin(x)

fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

Matplotlib graphs your data on <strong>Figure</strong>s (e.g. windows, Jupyter widgets, etc.), each of which can contain one or more <strong>Axes</strong>, an area where points can be specified in terms of x-y coordinates (or theta-r in a polar plot, x-y-z in a 3-D plot, etc.)

The simplest way to create a Figure with an Axes is using pyplot.subplots. We can then use Axes.plot to draw data on the Axes, then use <strong>show()</strong> to display the Figure:

```python
fig, ax = plt.subplots() # Create a figure containing a single Axes
ax.plot([1,2,3,4], [5,6,7,8]) # Plot some data on the Axes
plt.show() # Show the figure
# Note: depending on the environment you're working in,
# you can leave out plt.show(); for Jupyter notebooks, 
# it shows automatically.
```

Parts of a Figure:


<strong><em>Figure</em></strong>

The <strong>whole</strong> figure. The Figure keeps track of all the child Axes, a group of 'special' Artists (titles, figure legends, colorbars, and more), and even nested subfigures. 

Typically, you'll create a new Figure through one of the following functions:

```python
fig = plt.figure() # an empty figure with no Axes
plt.show()
```

```python
fig, ax = plt.subplots() # a figure with a single Axes
plt.show()
```

```python
fig, axs = plt.subplots(2,2) # a figure wirth a 2x2 grid of Axes
plt.show()
```

```python
# a figure with one Axes on the left, and two on the right
fig, axs = plt.subplot_mosaic([['left', 'right_top'],
                               ['left', 'right_bottom']])
```

subplots() and subplot_mosaic() are convinience functions that additionally create Axes objects inside the Figure, but you can manually add Axes later on. 

<strong><em>Axes</em></strong>

An Axes is an Artist attached to a Figure that contains a region for plotting data, and usually include two (or three, in the case of 3-D) <strong>Axis</strong> objects (be aware of the difference between Axes and Axis) that provide ticks and tick labels to provide scales for the data in the Axes. 

Each Axes also has a title (set via set_title()), an x-label (set via set_xlabel()), and a y-label (set via set_ylabel()).

The Axes methods are the primary interface for configurng most parts of your plot (adding data, controlling axis scales and limits, adding labels, etc.)

<strong><em>Axis</em></strong>

These objects set the scale, limits, and generate ticks (marks on the Axis on the graph) and ticklabels (strings labeling ticks). The location of the ticks is determined by a Locator object and the ticklabel strings are formatted by a Formatter. The combination of the correct Locator and Formatter gives very fine control over the tick locations and labels.

<strong><em>Artist</em></strong>

Basically, everything visible on the Figure is an Artist (even Figure, Axes, and Axis objects). This includes Text objects, Line2D objects, collections objects, Patch objects, etc. When the Figure is rendered, all of the Artists are drawn to the <strong>canvas</strong>.

 Most Artists are tied to an Axes; such an Artist cannot be shared by multiple Axes, or moved from one to another. 

<h2>Types of inputs to plotting functions</h2>

Plotting functions expect <strong>numpy.array</strong> or <strong>numpy.ma.masked_array</strong> as input, or objects that can be passed to numpy.asarray. Classes that are similar to arrays ('array-like') such as <strong>pandas</strong> data objects and <strong>numpy.matrix</strong> may not work as intended. Common convention is to convert these objects prior to plotting. For example, to convert a <strong>numpy.matrix</strong>:

```python
import numpy as np
b = np.matrix([[1,2], [3,4]])
b_asarray = np.asarray(b)
b
```

Most methods will also parse a string-indexable object like a <em>dict</em>, a structured numpy array, or a pandas.DataFrame. 

Matplotlib allows you to provide the data keyword argument and generate plots passing the strings corresponding to the x and y variables.

```python
np.random.seed(19680801) # seed the random number generator
data = {'a': np.arange(50),
        'c': np.random.randint(0, 50, 50),
        'd': np.random.randn(50)}
data['b'] = data['a'] + 10 * np.random.randn(50)
data['d'] = np.abs(data['d']) * 100

fig, ax = plt.subplots(figsize=(5, 2.7), layout='constrained')
ax.scatter('a','b',c='c', s='d', data=data)
ax.set_xlabel('entry a')
ax.set_ylabel('entry b')
```
