NumPy's main feature is the homogenous. multidimensional array. It is a table of elements (usually numbers) all of the same type, indexed by a tuple of non-negative integers.
-in NumPy, dimensions are called axes (plural of axis)

The array for coords of point in 3D space below has one axis. This axis has 3 elements, so it is an axis of length 3 (1 row by 3 columns):

[1,2,3]

The array below has 2 axes. The first axis has a length of 2, and the second axis has a length of 3 (2 rows by 3 columns):

[[1, 2, 3],
 [4, 5, 6]]

NumPy's array class is called ndarray, and has the alias array.
-numpy.array is not the same as the Standard Python Library's array.array; STL handles 1-D arrays with less functionality.

important numpy.array attributes:
-ndarray.ndim -> # of axes (dimensions) of the array

-ndarray.shape -> dimensions of the array. This is a tuple of integers indicating the size of the array in each dimension. For a matrix with n rows and m columns, shape will be (n,m). the length of the shape tuple, is therefore the # of axes, ndim

-ndarray.size -> the total number of elements of the array. This is equal to the product of the elements of shape.

-ndarray.dtype -> an object describing the type of elements in the array. One can create or specify dtypes using standard Python types. NumPy also provides its own types, such as numpy.int32, numpy.int16. and numpy.float64

-ndarray.itemsize -> the size in bytes of each element of the array. e.g., an array of elements of type float64 has itemsize 8 (=64/8), while one of type complex32 has itemszie 4 (=32/8). It's equal to ndarray.dtype.itemsize. (remember, a byte is 8 bits)

-ndarray.data -> the buffer containing the actual elements of the array. Normally, we won't need this attribute because we will access elements in an array using indexing facilitites.

ARRAY CREATION - there are several ways to create arrays, e.g. from a Python STL list / tuple using the array function.

 The type of resulting array is deduced from the type of elements in the sequences:

```python
a = np.array([2,3,4])
a
```

```python
a.dtype
```

```python
b = np.array([1.3, 4.5, 6.7])
b.dtype
```

```python
c = np.array(["one", "two", "three"])
c.dtype
```

A frequent error comes from calling np.array() with multiple argument, rather than providing a single sequence as an argument:

```python
a = np.array(1,2,3,4) # Wrong
```

```python
a = np.array([1,2,3,4]) # Correct
```

np.array transforms sequences of sequences into 2-D arrays, sequences of sequences of sequences to 3-D arrays, etc:

```python
b = np.array([(1,2,3,4), (5,6,7,8)])
b
```

```python
b.ndim
```
